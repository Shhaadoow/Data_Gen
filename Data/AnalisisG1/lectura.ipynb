```python
import os
from pathlib import Path

base = Path(
    r"C:\Users\ASUS\.cache\huggingface\hub\datasets--lighteternal--pgc-psychiatric-gwas-harmonized\snapshots\6c3241136562df51a4e942d84a466a4202866c99"
)

def obtener_bytes(ruta):
    return ruta.stat().st_size


print("SUMA TOTAL POR TRASTORNO")
print("=" * 50)

archivos_data = {
    archivo.name: archivo
    for archivo in (base / "data").glob("*.parquet")
}

archivos_analysis = {
    archivo.name: archivo
    for archivo in (base / "analysis").glob("*.parquet")
}

total_global_bytes = 0

for nombre in archivos_data:

    # Solo archivos presentes en ambas carpetas
    if nombre in archivos_analysis:

        bytes_data = obtener_bytes(archivos_data[nombre])
        bytes_analysis = obtener_bytes(archivos_analysis[nombre])

        total_bytes = bytes_data + bytes_analysis

        total_gb = total_bytes / (1024 ** 3)
        total_mb = total_bytes / (1024 ** 2)

        total_global_bytes += total_bytes

        print(f"{nombre}")
        print(f"Data: {bytes_data:,} bytes")
        print(f"Analysis: {bytes_analysis:,} bytes")
        print(f"Total: {total_bytes:,} bytes")
        print(f"Total MB: {total_mb:.5f} MB")
        print(f"Total GB: {total_gb:.5f} GB")
        print("-" * 50)


# =========================================
# TOTAL GLOBAL
# =========================================

total_global_mb = total_global_bytes / (1024 ** 2)
total_global_gb = total_global_bytes / (1024 ** 3)

print("\nTOTAL GLOBAL DATA + ANALYSIS")
print("=" * 50)

print(f"Total exacto bytes: {total_global_bytes:,}")
print(f"Total exacto MB: {total_global_mb:.5f} MB")
print(f"Total exacto GB: {total_global_gb:.5f} GB")
```

```python
from pathlib import Path
import polars as pl

# =========================================
# RUTA BASE DATASET
# =========================================

base = Path(
    r"C:\Users\ASUS\.cache\huggingface\hub\datasets--lighteternal--pgc-psychiatric-gwas-harmonized\snapshots\6c3241136562df51a4e942d84a466a4202866c99"
)

# =========================================
# CONFIGURACIÓN POLARS
# =========================================

pl.Config.set_tbl_cols(-1)
pl.Config.set_tbl_rows(10)
pl.Config.set_tbl_width_chars(300)
pl.Config.set_fmt_str_lengths(120)

# =========================================
# FUNCIÓN PRINCIPAL
# =========================================

def describir_parquet(ruta_archivo):

    print("\n")
    print("=" * 100)
    print(f"ARCHIVO: {ruta_archivo.name}")
    print("=" * 100)

    # =====================================
    # LECTURA LAZY
    # =====================================

    df = pl.scan_parquet(ruta_archivo)

    # =====================================
    # SCHEMA
    # =====================================

    schema = df.collect_schema()

    # =====================================
    # HEAD Y TAIL
    # =====================================

    head = df.head(2).collect()
    tail = df.tail(2).collect()

    # =====================================
    # TOTAL FILAS Y COLUMNAS
    # =====================================

    total_filas = df.select(pl.len()).collect().item()
    total_columnas = len(schema)

    print(f"\nFilas x Columnas: {total_filas:,} x {total_columnas}")

    # =====================================
    # CONSTRUCCIÓN TABLA DESCRIPTIVA
    # =====================================

    filas = []

    for columna, tipo in schema.items():

        fila = {
            "columna": columna,
            "tipo": str(tipo),

            "fila_1":
                str(head[columna][0])
                if len(head) > 0 else None,

            "fila_2":
                str(head[columna][1])
                if len(head) > 1 else None,

            "ultima_1":
                str(tail[columna][0])
                if len(tail) > 0 else None,

            "ultima_2":
                str(tail[columna][1])
                if len(tail) > 1 else None,
        }

        filas.append(fila)

    tabla = pl.DataFrame(filas)

    print("\nTABLA DESCRIPTIVA")
    print("-" * 100)

    print(tabla)


# =========================================
# RECORRER DATA
# =========================================

print("\n")
print("#" * 100)
print("CARPETA DATA")
print("#" * 100)

for archivo in (base / "data").glob("*.parquet"):

    describir_parquet(archivo)


# =========================================
# RECORRER ANALYSIS
# =========================================

print("\n")
print("#" * 100)
print("CARPETA ANALYSIS")
print("#" * 100)

for archivo in (base / "analysis").glob("*.parquet"):

    describir_parquet(archivo)
```

```python
from pathlib import Path
import polars as pl

# =========================================
# CONFIGURACIÓN POLARS
# =========================================

pl.Config.set_tbl_cols(-1)
pl.Config.set_tbl_rows(50)
pl.Config.set_tbl_width_chars(500)
pl.Config.set_fmt_str_lengths(150)

# =========================================
# RUTA BASE
# =========================================

base = Path(
    r"C:\Users\ASUS\.cache\huggingface\hub\datasets--lighteternal--pgc-psychiatric-gwas-harmonized\snapshots\6c3241136562df51a4e942d84a466a4202866c99"
)

# =========================================
# ARCHIVOS ADHD
# =========================================

archivos = [
    base / "data" / "autism.parquet",
    base / "analysis" / "autism.parquet"
]

# =========================================
# FUNCIÓN DE ANÁLISIS
# =========================================

def analizar_archivo(ruta):

    print("\n")
    print("=" * 120)
    print(f"ARCHIVO: {ruta}")
    print("=" * 120)

    # =====================================
    # LECTURA LAZY
    # =====================================

    df = pl.scan_parquet(ruta)

    # =====================================
    # FILAS Y COLUMNAS
    # =====================================

    total_filas = df.select(pl.len()).collect().item()
    total_columnas = len(df.collect_schema())

    print(f"\nFilas: {total_filas:,}")
    print(f"Columnas: {total_columnas}")

    # =====================================
    # PRIMERAS 2 FILAS
    # =====================================

    head = df.head(2).collect()

    print("\n")
    print("-" * 120)
    print("PRIMERAS 2 FILAS")
    print("-" * 120)

    print(head)

    # =====================================
    # ÚLTIMAS 2 FILAS
    # =====================================

    tail = df.tail(2).collect()

    print("\n")
    print("-" * 120)
    print("ÚLTIMAS 2 FILAS")
    print("-" * 120)

    print(tail)

    # =====================================
    # COLUMNAS Y TIPOS
    # =====================================

    print("\n")
    print("-" * 120)
    print("COLUMNAS Y TIPOS")
    print("-" * 120)

    schema = df.collect_schema()

    for columna, tipo in schema.items():
        print(f"{columna} --> {tipo}")


# =========================================
# EJECUCIÓN
# =========================================

for archivo in archivos:
    analizar_archivo(archivo)
```
