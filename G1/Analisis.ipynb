Las variantes genéticas son cambios en la secuencia de ADN. 
Para definir una alteración específica, se utilizan los siguientes conceptos clave:
    - SNP (Polimorfismo de un solo nucleótido): Es el tipo más común. Ocurre cuando se altera una sola letra (A, T, C o G) en el genoma (por ejemplo, tener una "A" en  lugar de una "G").
    - Inserciones/Deleciones (Indels): Son variantes donde se añade (inserción) o se elimina (deleción) uno o más pares de bases del ADN.
    - Consecuencias: Estas variantes pueden ser benignas, patogénicas (causantes de enfermedades) o de significado incierto.
