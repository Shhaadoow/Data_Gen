```python
import polars as pl
import pandas as pd
from pathlib import Path

# =========================================
# CONFIGURACIÓN VISUAL
# =========================================

pl.Config.set_tbl_cols(-1)
pl.Config.set_tbl_rows(10)
pl.Config.set_tbl_width_chars(300)
pl.Config.set_fmt_str_lengths(100)

# =========================================
# RUTA DATA/eating_disorders
# =========================================

ruta = Path(r"../Data/eating_disorders.parquet")

# =========================================
# LECTURA
# =========================================

df = pl.read_parquet(ruta)

# =========================================
# MOSTRAR FxC
# =========================================

print("=" * 100)
print("eating_disorders.PARQUET - DATA")
print("=" * 100)

df.select(pl.col("chr").n_unique())
# ver cuantos cromosomas participan, y cuantas variantes estan relacionadas a ese. 
df.group_by("chr").count().sort("count", descending=True)
# Filtrado por pvalue
df.select(
    (pl.col("pval") < 0.05).sum().alias("p_menor_005"),
    (pl.col("pval") < 5e-8).sum().alias("gwas_significativas"))


print("\nCROMOSOMAS PRESENTES\n")

print(
    df.group_by("chr")
      .len()
      .sort("len", descending=True)
)


print("\nNUMERO DE CROMOSOMAS\n")

print(
    df.select(
        pl.col("chr").n_unique().alias("cromosomas_unicos")
    )
)


gwas = df.filter(
    pl.col("pval") < 5e-8
)

print("\nGWAS SIGNIFICATIVAS POR CROMOSOMA\n")

print(
    gwas.group_by("chr")
        .len()
        .sort("len", descending=True))



```
