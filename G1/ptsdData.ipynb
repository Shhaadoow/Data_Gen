ANALISIS DEL TRASTORNO PTSD

```python
import polars as pl
import pandas as pd
from pathlib import Path

# =========================================
# CONFIGURACIÓN VISUAL
# =========================================

pl.Config.set_tbl_cols(-1)
pl.Config.set_tbl_rows(10)
pl.Config.set_tbl_width_chars(300)
pl.Config.set_fmt_str_lengths(100)

# =========================================
# RUTA DATA/PTSD
# =========================================

ruta = Path(r"../Data/ptsd.parquet")

# =========================================
# LECTURA
# =========================================

df = pl.read_parquet(ruta)

# =========================================
# MOSTRAR FxC
# =========================================

print("=" * 100)
print("PTSD.PARQUET - DATA")
print("=" * 100)

print(f"\nFilas x Columnas: {df.shape}")

# =========================================
# MOSTRAR 10 FILAS Y 31 COLUMNAS
# =========================================

print("\nPRIMERAS 10 FILAS\n")
print(df.shape[0])

df.select(pl.col("chr").n_unique())


# ver cuantos cromosomas participan, y cuantas variantes estan relacionadas a ese. 
df.group_by("chr").count().sort("count", descending=True)

# Filtrado por pvalue
df.select(
    (pl.col("pval") < 0.05).sum().alias("p_menor_005"),
    (pl.col("pval") < 5e-8).sum().alias("gwas_significativas"))

#Tabla de filtradas

resultado = (
    df.filter(
    pl.col("pval") < 5e-8
).select(
    [
        "snp_id",
        "chr",
        "pos",
        "info",
        "effect_size",
        "pval"
    ]
)
      .sort("pval")
)

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)
pd.set_option("display.max_colwidth", None)

display(resultado.to_pandas())
```

```python
import polars as pl
from pathlib import Path

# Primeras 5 filas
primeras = df.head(5)

# Últimas 5 filas
ultimas = df.tail(5)

# Unir ambas
muestra = pl.concat([primeras, ultimas])

# Guardar en Excel
salida = Path("../Resultados/ptsd_muestra.xlsx")
salida.parent.mkdir(exist_ok=True)

muestra.write_excel(salida)

print(f"Excel generado en: {salida.resolve()}")
```
